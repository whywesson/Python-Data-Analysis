# 项目：评估和清理英国电商公司销售数据

## 分析目标

此数据分析的目的是，根据市场销售数据，挖掘畅销产品，以便制定更有效的市场策略来提升营收。

本实战项目的目的在于练习评估数据干净和整洁度，并且基于评估结果，对数据进行清洗，从而得到可供下一步分析的数据。

## 简介

原始数据集记录了一家英国在线零售公司在2010年12月1日至2011年12月9日期间的所有交易情况，涵盖了该公司在全球不同国家和地区的业务数据。该公司主要销售覆盖各个场景的礼品，包括但不限于生日礼品、结婚纪念品、圣诞礼品等等。该公司的客户群体主要包括批发商和个人消费者，其中批发商占据了相当大的比例。

数据每列的含义如下：
- `InvoiceNo`: 发票号码。6位数，作为交易的唯一标识符。如果这个代码以字母“c”开头，表示这笔交易被取消。
- `StockCode`: 产品代码。5位数，作为产品的唯一标识符。
- `Description`: 产品名称。
- `Quantity`: 产品在交易中的数量。
- `InvoiceDate`: 发票日期和时间。交易发生的日期和时间。
- `UnitPrice`: 单价。价格单位为英镑（£）。
- `CustomerID`: 客户编号。5位数，作为客户的唯一标识符。
- `Country`: 国家名称。客户所居住的国家的名称。

## 读取数据

```python
import pandas as pd
import numpy as np
```

```python
original_data=pd.read_csv('e_commerce.csv')
original_data.head()
```

## 评估数据

### 评估数据结构整洁度

```python
original_data.sample(10)
```

从随机的10行数据中可以判断，每一行为每张发票中每种产品代码的的交易情况，每一列为每个交易情况的相关变量，结构上符合每行只有一个观察值，每列只有一个变量，每个单元格为一个观察值的整洁数据的特点。

### 评估数据内容干净度

评估数据内容的干净程度，主要从空缺值、重复数据、不一致数据、无效/错误数据等方面评估。

```python
original_data.info()
```

通过info方法得知，共有541909条观察值，其中Description、CustomerID存在空缺数据，InvoiceDate的数据类型应为日期时间，CustomerID数据类型应为字符串，需要做数据类型转换。

#### 评估空缺数据

将Description列空缺的行提取出来。

```python
original_data[original_data['Description'].isnull()]
```

共筛选出来1454行，观察发现，这些Description列为空值的观察值中，UnitPrice的值似乎都为0，为了验证猜测，进一步通过条件筛选，看看是否有Description值为空，且UnitPrice值不为0的行。

```python
original_data[(original_data['Description'].isnull()) & (original_data['UnitPrice']!=0)]
```

条件筛选结果为空，因此猜想成立，即Description列为空值的观察值，它们的UnitPrice的值都为0。

而此次分析的目的是挖掘畅销产品，而产品名称Description和产品单价UnitPrice都是对分析有重要意义的变量，因此当两者同时缺少或无效时，对后续分析无意义，因此需要将Description为空的行删除。

将CustomerID列空缺的行提取出来。

```python
original_data[original_data['CustomerID'].isnull()]
```

共提取出135080行CustomerID为空值的观察值。这些行除了CustomerID外，其他列都为有效数据，且CustomerID指代客户编号，空缺后对实际销售数据分析无影响，因此无需对CustomerID为空的行做处理。

#### 评估重复数据

从各列的意义来看，虽然InvoiceNo、StockCode、CustomerID均为交易、产品、客户的唯一标识符，但同一交易允许存在不同产品，同一产品可存在不同交易中，同一客户可一次下单不同产品或多次下单，因此这三个变量均可存在重复。

#### 评估不一致数据

不一致数据是指用不同的值指代同一目标，数据中Country变量有可能存在不一致数据。

```python
original_data['Country'].value_counts()
```

在Country列的所有值中，United Kingdom、UK、U.K.都指代英国，USA、United States都指代美国，因此需要将UK、U.K.统一替换成United Kingdom，将USA替换程United States。

#### 评估无效/错误数据

使用describe方法检查数值数据的整体描述。

```python
original_data.describe()
```

Quantity和UnitPrice的最小值均为负数，不符合实际情况，因此这两列含有无效或错误数据。

通过条件筛选出Quantity列无效/错误数据的行。

```python
original_data[original_data['Quantity']<0]
```

共有10624 rows行的Quantity变量值为负数。通过观察发现这些行似乎InvoiceNo都以“C”开头，表示取消订单，为验证猜想，需要通过条件筛选，检查是否有Quantity为负数，而InvoiceNo不以C开头的行。

```python
original_data[(original_data['InvoiceNo'].str[0]!='C')&(original_data['Quantity']<0)]
```

有1336行满足筛选条件，因此猜想不成立。但发现在这些行里面，UnitPrice的似乎值都为0，为验证猜想，需要进一步对其进行条件筛选，检查其中是否有UnitPrice不为0的行。

```python
original_data[(original_data['InvoiceNo'].str[0]!='C')&(original_data['Quantity']<0)&(original_data['UnitPrice']!=0)]
```

筛选结果为空，因此猜想成立，即Quantity为负数且InvoiceNo不以C开头的观察值中UnitPrice值全部为0。

因此Quantity为负数的情况有如下两种：
- InvoiceNo以C开头，表示订单取消
- UnitPrice为0，即单价为0英镑  

这些数据均不是有效成交数据，不贡献销售，因此对分析畅销产品无意义，所以后续数据清洗环节需要将Quantity为负数的数据删除。

通过条件筛选出UnitPrice列无效/错误数据的行。

```python
original_data[original_data['UnitPrice']<0]
```

共有2个UnitPrice值为负数的观察值，从Description的含义来看，为调整坏账，并非有效销售数据，对后续的销售情况分析无意义，因此需要在数据清洗环节删除。

## 清理数据

从前面的数据评估环节得知，需要将数据做如下清洗：
- 将Description列为空的行删除
- 将UK、U.K.统一替换成United Kingdom
- 将USA替换程United States
- 将Quantity为负数的数据删除
- 将UnitPrice为负数的数据删除
- 将InvoiceDate的数据类型转换为日期时间
- 将CustomerID的数据类型转换为字符串

在清洗之前，需要将原始数据复制给新的数据cleaned_data，这样后续所有的清洗步骤均操作在复制出来的副本cleaned_data上。

```python
cleaned_data = original_data.copy()
cleaned_data.head()
```

#### 删除Description列为空的行

删除Description列为空的行，并检查删除后该列空缺值的个数是否为0

```python
cleaned_data.dropna(subset=['Description'],inplace=True)
cleaned_data['Description'].isnull().sum()
```

#### 将UK、U.K.统一替换成United Kingdom

```python
cleaned_data['Country']=cleaned_data['Country'].replace(['UK','U.K.'],'United Kingdom')
```

```python
print(len(cleaned_data[cleaned_data['Country']=='UK']))
print(len(cleaned_data[cleaned_data['Country']=='U.K.']))
```

#### 将USA替换成United States

```python
cleaned_data['Country'] = cleaned_data['Country'].replace('USA','United States')
```

```python
print(len(cleaned_data[cleaned_data['Country']=='USA']))
```

#### 将Quantity为负数的数据删除

```python
cleaned_data = cleaned_data[cleaned_data['Quantity']>=0]
```

```python
len(cleaned_data[cleaned_data['Quantity']<0])
```

#### 将UnitPrice为负数的数据删除

```python
cleaned_data = cleaned_data[cleaned_data['UnitPrice']>=0]
```

```python
len(cleaned_data[cleaned_data['UnitPrice']<0])
```

#### 将InvoiceDate的数据类型转换为日期时间

```python
cleaned_data['InvoiceDate'] = pd.to_datetime(cleaned_data['InvoiceDate'])
cleaned_data['InvoiceDate']
```

#### 将CustomerID的数据类型转换为字符串

```python
cleaned_data['CustomerID'] = cleaned_data['CustomerID'].astype(str)
cleaned_data['CustomerID']
```

还需要去除结尾的.0

```python
cleaned_data['CustomerID'] = cleaned_data['CustomerID'].str.slice(0,-2)
cleaned_data['CustomerID']
```

清理后的数如下：

```python
cleaned_data.sample(10)
```

## 保存清理后的数据

完成数据清理后，将干净整洁的数据保存到文件名为e_commerce_cleaned.csv的新文件里。

```python
cleaned_data.to_csv('e_commerce_cleaned.csv', index=False)
```
