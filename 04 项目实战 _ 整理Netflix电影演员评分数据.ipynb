# 项目：整理Netflix电影演员评分数据

## 分析目标

此数据分析的目的是，整理不同流派影视作品，比如喜剧片、动作片、科幻片中，各演员出演作品的平均IMDB评分，从而挖掘出各个流派中的高评分作品演员。

本实战项目的目的在于练习整理数据，从而得到可供下一步分析的数据。

## 简介

原始数据集记录了截止至2022年7月美国地区可观看的所有Netflix电视剧及电影数据。数据集包含两个数据表：`titles.csv`和`credits.csv`。

`titles.csv`包含电影及电视剧相关信息，包括影视作品ID、标题、类型、描述、流派、IMDB（一个国外的在线评分网站）评分，等等。`credits.csv`包含超过7万名出现在Netflix影视作品的导演及演员信息，包括名字、影视作品ID、人物名、演职员类型（导演/演员）等。

`titles.csv`每列的含义如下：
- id：影视作品ID。
- title：影视作品标题。
- show_type：作品类型，电视节目或电影。
- description：简短描述。
- release_year：发布年份。
- age_certification：适龄认证。
- runtime：每集电视剧或电影的长度。
- genres：流派类型列表。
- production_countries：出品国家列表。
- seasons：如果是电视剧，则是季数。
- imdb_id：IMDB的ID。
- imdb_score：IMDB的评分。
- imdb_votes：IMDB的投票数。
- tmdb_popularity：TMDB的流行度。
- tmdb_score：TMDB的评分。

`credits.csv`每列的含义如下：
- person_ID：演职员ID。
- id：参与的影视作品ID。
- name：姓名。
- character_name：角色姓名。
- role：演职员类型，演员或导演。

## 数据整理

```python
cleaned_title_data.sample(10)
```

```python
cleaned_credits_data.sample(10)
```

由于该数据分析的目的是整理不同流派的各演员参演作品的IMDB平均评分，从而挖掘出各流派中评分最高的演员，而流派（genres）、IMDB评分（imdb_score）、演职员ID（person_id）分别属于不同的DataFrame,如果想要得到不同流派的各演员参演作品的IMDB平均评分，就需要将cleaned_title_data和cleaned_credits_data两个DataFrame通过共同的键id进行合并，我们将合并后的DataFrame取名为title_and_credits。

```python
title_and_credits = pd.merge(cleaned_title_data, cleaned_credits_data, on ='id')
title_and_credits.head()
```

在role变量中还含有DIRECTOR（导演）的值，由于是要整理各流派影视作品中各演员的IMDB评分，只包含演员并不包含导演，因此需要将role变量的值为DIRECTOR（导演）的列删除，并检查删除后数据title_and_credits中role的值为DIRECTOR的行数。

```python
title_and_credits = title_and_credits.query('role != "DIRECTOR"')
title_and_credits.query('role == "DIRECTOR"')
```

以上根据条件筛选出role值为DIRECTOR的DataFrame为空，说明已成功将role值为DIRECTOR的行删除。

接下来将title_and_credits按照流派（genres）和演员ID（person_id）为索引分组，提取imdb_score变量，计算出IMDB评分的平均值。

```python
score_groupby_genres_and_person = title_and_credits.groupby(['genres','person_id'])['imdb_score'].mean()
score_groupby_genres_and_person
```

调用reset_index方法，重置索引，这样会得到一个规范的DataFrame。

```python
score_groupby_genres_and_person = score_groupby_genres_and_person.reset_index()
score_groupby_genres_and_person
```

将重置索引得到的score_groupby_genres_and_person按照流派（genres）分组，提取imdb_score的最大值，得到各流派IMDB最大值的Series。

```python
max_score_groupby_genres = score_groupby_genres_and_person.groupby('genres')['imdb_score'].max()
max_score_groupby_genres
```

将max_score_groupby_genres与score_groupby_genres_and_person以genres和imdb_score为键进行内连接，这样就能筛选出各流派影视作品中最高分的演员ID（person_id）。

```python
max_score_groupby_genres_and_person = pd.merge(score_groupby_genres_and_person, max_score_groupby_genres, on=['genres', 'imdb_score'], how='inner')
max_score_groupby_genres_and_person
```

这样得到了各流派影视作品中IMDB评分最高的演员ID，其中各个流派中的最高分演员可能不止一位。

接下来还需要知道各流派作品中评分最高的演员ID对应的名字，因此需要将max_score_groupby_genres_and_person和cleaned_credits_data进行连接，因为cleaned_credits_data还有其他列，而我们只需要用到person_id和name这两列，因此先提取这两列，再删除重复行。

```python
person_and_name = cleaned_credits_data[['person_id','name']].drop_duplicates()
person_and_name
```

将max_score_groupby_genres_and_person与person_and_name以person_id为键进行连接，得到包含各流派最高评分作品演员ID对应的演员姓名。

```python
max_score_actor_goupby_genres = pd.merge(max_score_groupby_genres_and_person, person_and_name, on = 'person_id')
max_score_actor_goupby_genres
```
